```python
print("="*60)
print("           EMPLOYEE SALARY MANAGEMENT SYSTEM ")
print("="*60)

emp_id = input("Enter Employee ID :")
name = input("Enter Employee Nmae :")
department = input("Enter Department :")
salary = int(input("Enter Salary :"))
experience = int(input("Enter Experience years :"))
attendance = int(input("Enter Attendance(%) :"))
rating = int(input("Enter Performance Rating(1-5) :"))

# Bonus
if salary>=80000:
    bonus = salary*0.15
elif salary>=60000:
    bonus = salary*0.10
elif salary>=40000:
    bonus = salary*0.07
else:
    bonus = salary*0.05

#HRA
if salary>=60000:
    hra=salary*0.02
else:
    hra=salary*0.10

# Tax
if salary >= 100000:
    tax = salary * 0.15
elif salary >= 70000:
    tax = salary * 0.10
elif salary >= 40000:
    tax = salary * 0.05
else:
    tax = 0

net_salary = salary + bonus + hra - tax
# Attendance Status
if attendance >= 95:
    attendance_status = "Excellent"
elif attendance >= 85:
    attendance_status = "Good"
elif attendance >= 75:
    attendance_status = "Average"
else:
    attendance_status = "Poor"

# Performance
if rating == 5:
    performance = "Outstanding"
elif rating == 4:
    performance = "Excellent"
elif rating == 3:
    performance = "Good"
elif rating == 2:
    performance = "Average"
else:
    performance = "Needs Improvement"

# Promotion
if experience >= 5 and rating >= 4:
    promotion = "Eligible"
else:
    promotion = "Not Eligible"

# Salary Grade
if net_salary >= 100000:
    grade = "A+"
elif net_salary >= 80000:
    grade = "A"
elif net_salary >= 60000:
    grade = "B"
elif net_salary >= 40000:
    grade = "C"
else:
    grade = "D"

print("\n" + "=" * 60)
print("            EMPLOYEE SALARY REPORT")
print("=" * 60)

print("Employee ID        :", emp_id)
print("Employee Name      :", name)
print("Department         :", department)

print("-" * 60)

print("Basic Salary       :", salary)
print("Bonus              :", round(bonus, 2))
print("HRA                :", round(hra, 2))
print("Tax                :", round(tax, 2))
print("Net Salary         :", round(net_salary, 2))

print("-" * 60)

print("Attendance         :", attendance, "%")
print("Attendance Status  :", attendance_status)
print("Performance        :", performance)
print("Promotion          :", promotion)
print("Salary Grade       :", grade)

print("-" * 60)

if rating == 5:
    print("Message : Employee of the Month ⭐")
elif rating == 4:
    print("Message : Excellent Work! Keep it up.")
elif rating == 3:
    print("Message : Good Performance.")
elif rating == 2:
    print("Message : Improve your performance.")
else:
    print("Message : Immediate Improvement Required.")

print("=" * 60)
print("        THANK YOU")
print("=" * 60)                 
```
