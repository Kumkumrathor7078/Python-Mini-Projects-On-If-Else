```python
print("=" * 60)
print("           SMART ATM BANKING SYSTEM")
print("=" * 60)

name = input("Enter Customer Name : ")
account = input("Enter Account Number : ")

pin = int(input("Enter 4-Digit ATM PIN : "))

balance = 25000

if pin == 1234:

    print("\nLogin Successful")
    print("\n1. Check Balance")
    print("2. Deposit Money")
    print("3. Withdraw Money")
    print("4. Fast Cash")
    print("5. Mobile Recharge")
    print("6. Change PIN")
    print("7. Exit")

    choice = int(input("\nEnter Your Choice : "))

    if choice == 1:

        print("\nCurrent Balance :", balance)

        if balance >= 50000:
            print("Premium Customer")
        elif balance < 1000:
            print("Low Balance Warning")

    elif choice == 2:

        amount = int(input("Enter Deposit Amount : "))

        if amount > 0:
            balance = balance + amount
            print("Money Deposited Successfully")
            print("Updated Balance :", balance)
        else:
            print("Invalid Deposit Amount")

    elif choice == 3:

        amount = int(input("Enter Withdraw Amount : "))

        if amount <= 0:
            print("Invalid Amount")

        elif amount > balance:
            print("Insufficient Balance")

        else:
            balance = balance - amount

            print("Please Collect Your Cash")
            print("Remaining Balance :", balance)

            if balance < 1000:
                print("Warning! Your Balance is Low.")

    elif choice == 4:

        print("\nFast Cash Options")
        print("1. 500")
        print("2. 1000")
        print("3. 2000")

        option = int(input("Choose Option : "))

        if option == 1:
            amount = 500
        elif option == 2:
            amount = 1000
        elif option == 3:
            amount = 2000
        else:
            amount = 0

        if amount == 0:
            print("Invalid Option")

        elif amount > balance:
            print("Insufficient Balance")

        else:
            balance = balance - amount
            print("Please Collect Your Cash")
            print("Remaining Balance :", balance)

    elif choice == 5:

        mobile = input("Enter Mobile Number : ")
        recharge = int(input("Enter Recharge Amount : "))

        if recharge <= 0:
            print("Invalid Recharge Amount")

        elif recharge > balance:
            print("Insufficient Balance")

        else:
            balance = balance - recharge
            print("Recharge Successful")
            print("Mobile Number :", mobile)
            print("Remaining Balance :", balance)

    elif choice == 6:

        old_pin = int(input("Enter Old PIN : "))

        if old_pin == pin:
            new_pin = int(input("Enter New PIN : "))
            pin = new_pin
            print("PIN Changed Successfully")
        else:
            print("Wrong Old PIN")

    elif choice == 7:

        print("Thank You For Using Our ATM")

    else:

        print("Invalid Choice")

    print("\n")
    print("=" * 60)
    print("ATM RECEIPT")
    print("=" * 60)
    print("Customer Name :", name)
    print("Account Number :", account)
    print("Available Balance :", balance)
    print("=" * 60)

else:

    print("\nIncorrect PIN")
    print("Access Denied")
```
