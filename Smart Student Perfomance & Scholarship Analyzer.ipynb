```python
print("=" * 60)
print("      SMART STUDENT PERFORMANCE & SCHOLARSHIP ANALYZER")
print("=" * 60)

name = input("Enter Student Name : ")
roll = input("Enter Roll Number : ")

print("\nEnter Marks (Out of 100)\n")

english = int(input("English : "))
maths = int(input("Mathematics : "))
science = int(input("Science : "))
computer = int(input("Computer : "))
hindi = int(input("Hindi : "))

attendance = int(input("\nEnter Attendance (%) : "))

total = english + maths + science + computer + hindi
percentage = total / 5

# Result
if english < 33 or maths < 33 or science < 33 or computer < 33 or hindi < 33:
    result = "FAIL"
else:
    result = "PASS"

# Grade
if result == "FAIL":
    grade = "F"
elif percentage >= 90:
    grade = "A+"
elif percentage >= 80:
    grade = "A"
elif percentage >= 70:
    grade = "B"
elif percentage >= 60:
    grade = "C"
elif percentage >= 50:
    grade = "D"
else:
    grade = "E"

# Division
if result == "FAIL":
    division = "-"
elif percentage >= 60:
    division = "First Division"
elif percentage >= 45:
    division = "Second Division"
else:
    division = "Third Division"

# Scholarship
if percentage >= 90 and attendance >= 90 and result == "PASS":
    scholarship = "Eligible"
else:
    scholarship = "Not Eligible"

# Award
if result == "FAIL":
    award = "No Award"
elif percentage >= 95:
    award = "Gold Medal Candidate"
elif percentage >= 90:
    award = "Excellent Student Award"
elif percentage >= 80:
    award = "Merit Certificate"
else:
    award = "No Award"

# Attendance Status
if attendance >= 90:
    attendance_status = "Excellent"
elif attendance >= 75:
    attendance_status = "Good"
else:
    attendance_status = "Low Attendance"

# Parent Meeting
if attendance < 75 or result == "FAIL":
    parent_meeting = "Required"
else:
    parent_meeting = "Not Required"

# Performance
if result == "FAIL":
    performance = "Needs Improvement"
elif percentage >= 90:
    performance = "Outstanding"
elif percentage >= 80:
    performance = "Excellent"
elif percentage >= 70:
    performance = "Very Good"
elif percentage >= 60:
    performance = "Good"
else:
    performance = "Average"

print("\n")
print("=" * 60)
print("             STUDENT REPORT")
print("=" * 60)

print("Student Name      :", name)
print("Roll Number       :", roll)

print("-" * 55)

print("English           :", english)
print("Mathematics       :", maths)
print("Science           :", science)
print("Computer          :", computer)
print("Hindi             :", hindi)

print("-" * 55)

print("Total Marks       :", total, "/500")
print("Percentage        :", round(percentage, 2), "%")
print("Attendance        :", attendance, "%")
print("Attendance Status :", attendance_status)

print("-" * 55)

print("Result            :", result)
print("Grade             :", grade)
print("Division          :", division)
print("Scholarship       :", scholarship)
print("Award             :", award)
print("Performance       :", performance)
print("Parent Meeting    :", parent_meeting)

print("-" * 55)

if result == "PASS":
    if percentage >= 90:
        print("Suggestion : Excellent! Keep preparing for placements.")
    elif percentage >= 75:
        print("Suggestion : Good work. Practice daily to improve.")
    else:
        print("Suggestion : Work harder to score better.")
else:
    print("Suggestion : Focus on weak subjects and practice regularly.")

print("=" * 55)



```
